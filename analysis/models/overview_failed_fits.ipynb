```python
from pathlib import Path
import pandas as pd

def parse_file(path, town, phase_cut_date, prev_phase_cut_date=None):
    a, b = Path(path).read_text(encoding="utf-8").strip().splitlines()
    dropped = int(a.split()[1])
    total = int(a.split()[3])
    members = int(b.split()[1])
    return pd.DataFrame([{
        "dropped_rows": dropped,
        "total_rows": total,
        "considered_members": members,
        "town": town,
        "phase_cut_date": phase_cut_date,
        "prev_phase_cut_date": prev_phase_cut_date
    }])
```

## Bonn

```python
phase_cut_dates = ["2024-01-03", "2024-01-31", "2023-11-08", "2023-07-19", "2023-03-15"]
```

```python
town = "Bonn"
phase_cut_date = phase_cut_dates[0]


df_res = pd.DataFrame()
for phase_cut_date in phase_cut_dates:
    path = f"/home/iru-mls/marvin/ww_bonn_jax/{town}/multistart_results/{phase_cut_date}_cases_and_conc/visualizations_0.05/ensemble_members.txt"
    df_sub = parse_file(path, town, phase_cut_date)
    df_res = pd.concat([df_res, df_sub], ignore_index=True)
```
